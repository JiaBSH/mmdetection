```python
import json, os, sys

base = "/data/run01/scvi576/JiaBSH/mmdetection_para/dataset_root/mmdata_isat_1024"

# Write results to a file instead of stdout
output_file = "/data/home/scvi576/run/JiaBSH/mmdetection_para/dataset_counts.txt"

with open(output_file, 'w') as out:
    out.write(f"Using base: {base}\n")
    out.write(f"Exists: {os.path.exists(base)}\n")
    for split in ['train', 'val', 'test']:
        path = os.path.join(base, 'annotations', f'instances_{split}.json')
        if os.path.exists(path):
            with open(path) as f:
                d = json.load(f)
            out.write(f"{split}: images={len(d['images'])}, annotations={len(d['annotations'])}\n")
        else:
            out.write(f"{split}: NOT FOUND at {path}\n")

print(f"Results written to {output_file}")
print("Done")
```
